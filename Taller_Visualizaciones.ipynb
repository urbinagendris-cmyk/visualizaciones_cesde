## 1. Preparación
Elija **una** ruta de carga: CSV o SQLite. Ambas crean un DataFrame llamado `df`, de modo que el resto del cuaderno funciona sin cambios.

```python
from pathlib import Path
import sqlite3
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

sns.set_theme(style='whitegrid', palette='colorblind')
BASE = Path.cwd()
RUTA = BASE / 'data'
GRAFICOS = BASE / 'graficos'
GRAFICOS.mkdir(parents=True, exist_ok=True)
```

### Ruta A. CSV
Ejecute esta celda si desea trabajar con el archivo analítico procesado.

```python
df = pd.read_csv(RUTA / 'processed' / 'dataset_analitico.csv', parse_dates=['fecha'])
print(df.shape)
df.head()
```

### Ruta B. SQLite
Ejecute esta celda en lugar de la anterior si desea practicar consultas SQL. SQLite no necesita servidor.

```python
with sqlite3.connect(BASE / 'database' / 'centro_formacion.db') as conexion:
    tablas = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name", conexion)
    df = pd.read_sql_query('SELECT * FROM dataset_analitico', conexion)

df['fecha'] = pd.to_datetime(df['fecha'])
print(tablas)
print(df.shape)
df.head()
```

## 2. Reconocer los datos

```python
df.info()
display(df.describe(include='all').T)
```

## Comparación con barras

```python
df['curso'].value_counts().head(8).plot.bar(color='#17365D')
plt.title('Inscripciones por curso')
plt.xlabel('')
plt.ylabel('Inscripciones')
plt.xticks(rotation=35, ha='right')
plt.tight_layout()
plt.show()
```

## Evolución con líneas

```python
mensual = df.groupby(df['fecha'].dt.to_period('M')).size()
mensual.index = mensual.index.astype(str)
mensual.plot(marker='o', color='#E91E75')
plt.title('Inscripciones por mes')
plt.ylabel('Inscripciones')
plt.show()
```

## Distribución con histograma

```python
sns.histplot(data=df, x='nota', bins=12, kde=True, color='#17365D')
plt.title('Distribución de notas')
plt.show()
```

## Comparar distribuciones

```python
sns.boxplot(data=df, x='modalidad', y='nota')
plt.title('Notas por modalidad')
plt.show()
```

## Relación entre variables

```python
sns.scatterplot(data=df, x='progreso', y='nota', hue='estado', alpha=.75)
plt.title('Progreso y nota')
plt.show()
```

## Correlaciones

```python
variables = ['edad','nivel_socioeconomico','progreso','nota','satisfaccion','precio']
sns.heatmap(df[variables].corr(), annot=True, fmt='.2f', cmap='vlag', center=0)
plt.title('Matriz de correlaciones')
plt.show()
```
